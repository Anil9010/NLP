```python
import string
import random
import nltk

from nltk.corpus import stopwords, reuters
from collections import Counter, defaultdict
from nltk import FreqDist, ngrams

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('reuters')
nltk.download('punkt_tab')
sents = reuters.sents()

stop_words = set(stopwords.words('english'))


extra_removal_items = {'', '\t', 'rt', '-', '""', '"'}
removal_set = stop_words.union(set(list(string.punctuation) + list(extra_removal_items)))

unigram = []
bigram = []
trigram = []

for sentence_tokens in sents:
  cleaned_sentence_tokens = []
  for token in sentence_tokens:
    lower_token = token.lower()
    if lower_token not in removal_set:
      cleaned_sentence_tokens.append(lower_token)

  if cleaned_sentence_tokens:
    unigram.extend(cleaned_sentence_tokens)
    bigram.extend(list(ngrams(cleaned_sentence_tokens, 2, pad_left=True, pad_right=True)))
    trigram.extend(list(ngrams(cleaned_sentence_tokens, 3, pad_left=True, pad_right=True)))

freq_uni = FreqDist(unigram)
freq_bi = FreqDist(bigram)
freq_tri = FreqDist(trigram)

d = defaultdict(Counter)

for (a, b, c), frequency in freq_tri.items():
    if a is not None and b is not None and c is not None:
        d[(a, b)][c] += frequency


d_bigram = defaultdict(Counter)
for (a, b), frequency in freq_bi.items():
    if a is not None and b is not None:
        d_bigram[a][b] += frequency

def pick_word(counter):
  "Choose a random element"
  return random.choice(list(counter.elements()))

if not d:
    print("Trigram dictionary 'd' is empty. Cannot generate text.")
else:
    prefix = random.choice(list(d.keys()))

    print("Starting generation with prefix:", " ".join(prefix))

    if prefix in d:
        print(f"Debug: Starting text generation with prefix {prefix}")
        print(f"Debug: Trigram candidates for {prefix}: {d[prefix].most_common(5)}")

    generated_text = list(prefix)

    print(" ".join(generated_text))

    for i in range(19):
        next_word_candidates_trigram = list(d[prefix].elements())
        suffix = None

        if next_word_candidates_trigram:
            suffix = pick_word(d[prefix])
        else:
            last_word_of_prefix = prefix[1]
            if last_word_of_prefix in d_bigram and d_bigram[last_word_of_prefix]:
                suffix = pick_word(d_bigram[last_word_of_prefix])
            else:
                print(f"Generation stopped: No trigram or bigram found for prefix {prefix}")
                break

        generated_text.append(suffix)

        print(" ".join(generated_text))

        prefix = (prefix[1], suffix)


    output = ' '.join(generated_text)
```
