```python
text = input("Enter a sentence: ").lower()

if any(word in text.split() for word in ["happy", "joy", "excited", "great"]):
    emotion = "Happiness"
elif any(word in text.split() for word in ["sad", "unhappy", "lonely", "cry"]):
    emotion = "Sadness"
elif any(word in text.split() for word in ["angry", "hate", "furious", "annoyed"]):
    emotion = "Anger"
elif any(word in text.split() for word in ["afraid", "scared", "fear", "nervous"]):
    emotion = "Fear"
else:
    emotion = "Neutral"

print("Detected Emotion:", emotion)
```
